```python
path = kagglehub.dataset_download("arshid/iris-flower-dataset")
iris = pd.read_csv(path + "/IRIS.csv")
```

```python
from sklearn.cluster import KMeans
sse = []
x = iris.iloc[:, [0, 1, 2, 3]].values
for i in range(1, 9):
  kmeans = KMeans(n_clusters=i, init='random', max_iter=100, n_init=10, random_state = 2)
  kmeans.fit(x)
  sse.append(kmeans.inertia_)
```

```python
plt.plot(range(1, 9), sse)
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('SSE')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, init='random', max_iter=300, n_init=10, random_state = 2)
y_kmeans = kmeans.fit_predict(x)
```

```python
plt.scatter(x[y_kmeans == 0, 2], x[y_kmeans == 0, 3], s=100, c='purple', label = 'Iris-set')
plt.scatter(x[y_kmeans == 1, 2], x[y_kmeans == 1, 3], s=100, c='orange', label = 'Iris-ver')
plt.scatter(x[y_kmeans == 2, 2], x[y_kmeans == 2, 3], s=100, c='green', label = 'Iris-vir')

plt.scatter(kmeans.cluster_centers_[:, 2], kmeans.cluster_centers_[:, 3], s=100, c='red', label = 'Centroids')
plt.legend()
```

```python
from sklearn.metrics import silhouette_score
silhouette_scores = []

for k in range(2, 9):
  kmeans = KMeans(n_clusters=k, init='random', max_iter=100, n_init=10, random_state=2)
  cluster_labels = kmeans.fit_predict(x)
  score = silhouette_score(x, cluster_labels)
  silhouette_scores.append(score)

for i, score in enumerate(silhouette_scores, start=2):
  print(f"k ={i}, Score: {score}")
```

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.cluster import SpectralClustering

## Construim graful KNN
k = 5
knn = NearestNeighbors(n_neighbors=k)
knn.fit(x)
distances, indices = knn.kneighbors(x)
```

```python
## Creaam matricea de adiacenta
adj_matrix = np.zeros((x.shape[0], x.shape[0]))
for i in range(x.shape[0]):
  for j in indices[i]:
    adj_matrix[i, j] = 1

## Aplicam Spectral Clustering
num_clusters = 2
spectral = SpectralClustering(n_clusters=num_clusters, affinity='precomputed', random_state=42)
cluster_labels = spectral.fit_predict(adj_matrix)
```

## DBSCAN (Density-Based Spatial Clustering)

Aplicam DBSCAN pe acelasi set de date (`x` - cele 4 features din Iris), urmand aceeasi structura ca la K-Means si Spectral Clustering: estimam parametrii, aplicam algoritmul, vizualizam clusterele si evaluam rezultatul.

```python
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors

## Estimam eps folosind graficul k-distance
## Regula uzuala: min_samples = 2 * numarul de dimensiuni = 2 * 4 = 8
min_samples = 8

neigh = NearestNeighbors(n_neighbors=min_samples)
neigh.fit(x)
distances, indices = neigh.kneighbors(x)

## Luam distanta pana la al k-lea vecin si o sortam crescator
k_distances = np.sort(distances[:, min_samples - 1])

plt.plot(k_distances)
plt.title('K-distance graph (pentru alegerea eps)')
plt.xlabel('Puncte sortate')
plt.ylabel(f'Distanta pana la al {min_samples}-lea vecin')
plt.grid(True)
plt.show()
```

Punctul de "cot" (knee) al graficului de mai sus indica o valoare buna pentru `eps` - acolo unde distanta incepe sa creasca brusc. Pentru setul nostru, acesta se afla in jurul valorii `0.7`.

```python
## Aplicam DBSCAN
eps = 0.7
dbscan = DBSCAN(eps=eps, min_samples=min_samples)
dbscan_labels = dbscan.fit_predict(x)

## Eticheta -1 inseamna zgomot (noise / outlier)
dbscan_labels
```

```python
## Vizualizam clusterele pe aceleasi features ca la K-Means / Spectral (petal_length, petal_width)
plt.scatter(x[:, 2], x[:, 3], c=dbscan_labels, cmap='viridis', marker='o')
plt.xlabel('Petal length')
plt.ylabel('Petal width')
plt.title('DBSCAN clustering')
plt.colorbar(label='Cluster (-1 = zgomot)')
plt.show()
```

```python
## Distributia punctelor pe clustere (-1 = zgomot)
unique, counts = np.unique(dbscan_labels, return_counts=True)
print(unique, counts)

n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
n_noise = list(dbscan_labels).count(-1)
print(f'Numar de clustere gasite: {n_clusters}')
print(f'Numar de puncte de zgomot: {n_noise}')
```

```python
## Evaluam cu silhouette score (ca la K-Means)
## Excludem punctele de zgomot (-1) din calcul
from sklearn.metrics import silhouette_score

mask = dbscan_labels != -1
if len(set(dbscan_labels[mask])) >= 2:
  score = silhouette_score(x[mask], dbscan_labels[mask])
  print(f'Silhouette score (fara zgomot): {score}')
else:
  print('Nu exista suficiente clustere pentru silhouette score')
```

```python
## Testam mai multe valori pentru eps, ca sa vedem cum afecteaza rezultatul
for eps_val in [0.4, 0.5, 0.6, 0.7, 0.8, 0.9]:
  db = DBSCAN(eps=eps_val, min_samples=min_samples).fit(x)
  lbl = db.labels_
  n_cl = len(set(lbl)) - (1 if -1 in lbl else 0)
  n_ns = list(lbl).count(-1)
  m = lbl != -1
  if len(set(lbl[m])) >= 2:
    sc = silhouette_score(x[m], lbl[m])
    print(f'eps={eps_val}: clustere={n_cl}, zgomot={n_ns}, silhouette={sc:.3f}')
  else:
    print(f'eps={eps_val}: clustere={n_cl}, zgomot={n_ns}, silhouette=N/A')
```
